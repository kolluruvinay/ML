```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

print(f"TensorFlow Version: {tf.__version__}")

# 1. Load a standard dataset (IMDb movie review dataset)
# num_words specifies the maximum number of most frequent words to keep
num_words = 10000
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=num_words)

print(f"Training sequences: {len(x_train)}")
print(f"Test sequences: {len(x_test)}")



```

```python
# 2. Preprocess the data (padding sequences)
# Max sequence length. Reviews longer than this will be truncated, shorter will be padded.
maxlen = 256

x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=maxlen)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=maxlen)

print(f"Padded training data shape: {x_train.shape}")
print(f"Padded test data shape: {x_test.shape}")
```

```python
# 3. Build an RNN model
embedding_dim = 32
rnn_units = 32

model = keras.Sequential([
    layers.Embedding(num_words, embedding_dim),
    layers.SimpleRNN(rnn_units),
    layers.Dense(1, activation='sigmoid') # Sigmoid for binary classification
])

# Explicitly build the model to show parameters in summary
model.build(input_shape=(None, maxlen))

model.summary()
```

```python
# 4. Compile the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])


```

```python
# 5. Train the model
epochs = 5
batch_size = 128

history = model.fit(x_train, y_train,
                    epochs=epochs,
                    batch_size=batch_size,
                    validation_split=0.2) # Use 20% of training data for validation
```

```python
# 6. Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)
print(f"\nTest Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

print(f"TensorFlow Version: {tf.__version__}")

# 1. Load a standard dataset (IMDb movie review dataset)
# num_words specifies the maximum number of most frequent words to keep
num_words = 10000
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=num_words)

print(f"Training sequences: {len(x_train)}")
print(f"Test sequences: {len(x_test)}")



```

```python
# 2. Preprocess the data (padding sequences)
# Max sequence length. Reviews longer than this will be truncated, shorter will be padded.
maxlen = 256

x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=maxlen)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=maxlen)

print(f"Padded training data shape: {x_train.shape}")
print(f"Padded test data shape: {x_test.shape}")
```

```python
# 3. Build an LSTM model
embedding_dim = 32
lstm_units = 32

model_lstm = keras.Sequential([
    layers.Embedding(num_words, embedding_dim),
    layers.LSTM(lstm_units),
    layers.Dense(1, activation='sigmoid') # Sigmoid for binary classification
])

# Explicitly build the model to show parameters in summary
model_lstm.build(input_shape=(None, maxlen))

model_lstm.summary()



```

```python
# 4. Compile the LSTM model
model_lstm.compile(optimizer='adam',
                   loss='binary_crossentropy',
                   metrics=['accuracy'])
```

```python
# 5. Train the LSTM model
epochs = 10
batch_size = 128

history_lstm = model_lstm.fit(x_train, y_train,
                              epochs=epochs,
                              batch_size=batch_size,
                              validation_split=0.2) # Use 20% of training data for validation

```

```python
# 6. Evaluate the LSTM model
loss_lstm, accuracy_lstm = model_lstm.evaluate(x_test, y_test)
print(f"\nLSTM Test Loss: {loss_lstm:.4f}")
print(f"LSTM Test Accuracy: {accuracy_lstm:.4f}")
```
